# PROJEK BELAJAR MACHINE LEARNING - DETEKSI SPAM

```python
# DETEKSI SPAM
%pip install scikit-learn -q

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

#DATA - Semakin banyak data, semakin pintar
chat = [
    # SPAM
    "selamat anda menang hadiah 10 juta klik link ini",
    "pinjaman online cair cepat tanpa jaminan bunga rendah",
    "gratis iphone 15 cuma isi data diri anda",
    "akun anda terblokir verifikasi sekarang klik disini",
    "dapatkan saldo dana gratis 500rb",
    "lowongan kerja gaji 15jt tanpa pengalaman hubungi wa ini",
    "anda dapat voucher belanja gratis",
    "transfer sekarang untuk klaim hadiah anda",
    "promo pulsa murah 100rb cuma bayar 10rb",
    "selamat nomor anda terpilih dapat motor gratis",
    "investasi modal 100rb untung 10jt per hari",
    "cek bansos anda cair klik link berikut",
    "kirim kode otp anda untuk verifikasi hadiah",
    "diskon 90% semua produk hanya hari ini",
    "anda menang undian berhadiah mobil",
    "klaim hadiah uang tunai sekarang sebelum hangus",
    "akun bank anda bermasalah masukkan pin melalui link ini",
    "kerja dari rumah gaji jutaan daftar lewat whatsapp",
    "klik tautan ini untuk mendapatkan kuota internet gratis",
    "pemenang giveaway harap bayar biaya administrasi",
    "deposit sekarang profit investasi dijamin setiap hari",
    "paket anda tertahan bayar ongkir melalui link berikut",
    "bonus saldo gratis hanya untuk nomor terpilih",

    # BUKAN SPAM
    "bro besok jadi futsal ga jam 4",
    "tugas mtk udah dikumpul belum",
    "otw rumah lu 10 menit lagi",
    "bu nanti jemput jam berapa",
    "pinjem duit 20k dong besok gue ganti",
    "eh film yang kemarin seru banget",
    "besok bawa buku paket ya jangan lupa",
    "makan siang dimana kita",
    "nilai ulangan udah keluar cek di grup",
    "nanti malem mabar ga",
    "kirim foto catetan yang tadi dong",
    "hujan deres disini lu udah pulang belum",
    "besok ada upacara pake baju putih",
    "gue udah di depan gerbang",
    "jangan lupa bawa bola",
    "nanti sore kita belajar bareng di perpustakaan",
    "aku sudah kirim tugasnya ke email kamu",
    "tolong belikan air mineral saat pulang",
    "rapat kelompok mulai jam dua siang",
    "ibu minta kita mampir ke rumah nenek",
    "file presentasinya ada di folder bersama",
    "aku terlambat sedikit karena jalan macet",
    "jadi makan di kantin atau di luar",
]

label = ["spam"]*23 + ["bukan"]*23

#BAGI DATA - 80% buat belajar, 20% buat ujian
x_train, x_test, y_train, y_test = train_test_split(
    chat, label, test_size=0.2, random_state=42, stratify=label
)

#UBAH KATA JADI ANGKA
vector = TfidfVectorizer()
x_train_angka = vector.fit_transform(x_train)
x_test_angka = vector.transform(x_test)

#LATIH MODEL
model = MultinomialNB()
model.fit(x_train_angka, y_train)

#CEK AKURASI
hasil = model.predict(x_test_angka)
print(f"Akurasi model lu: {accuracy_score(y_test, hasil)*100:.0f}%")
print("---")

#COBA SENDIRI - Ganti-ganti isi chat disini
def cek_spam(teks):
    angka = vector.transform([teks])
    prediksi = model.predict(angka)[0]
    prob = model.predict_proba(angka)[0].max() * 100
    print(f"Chat: {teks}")
    print(f"Hasil: {prediksi.upper()} - yakin {prob:.0f}%")
    print("---")

cek_spam("SELAMAT ANDA MENANG HADIAH GRATIS KLIK LINK")
cek_spam("eh besok bawa buku paket ya jangan lupa")
cek_spam("pinjem duit 20k dong besok gue ganti")
cek_spam("Dilms ya, kalau disini saya gk tanggung jawab kalau tidak ternilai, karena bisa hilang fileny.")
```

## Ringkasan Dataset

Periksa jumlah pesan pada setiap kelas sebelum mengevaluasi model.

```python
from collections import Counter

jumlah_label = Counter(label)
print(f"Jumlah seluruh pesan: {len(chat)}")
for kategori, jumlah in jumlah_label.items():
    print(f"{kategori}: {jumlah} pesan")
```

## Evaluasi Model

Laporan klasifikasi menampilkan precision, recall, dan F1-score. Matriks kebingungan menggunakan urutan kelas `bukan`, lalu `spam`.

```python
print("Laporan klasifikasi:")
print(classification_report(y_test, hasil, labels=["bukan", "spam"], zero_division=0))

matriks = confusion_matrix(y_test, hasil, labels=["bukan", "spam"])
print("Matriks kebingungan (baris = label asli, kolom = prediksi):")
print("                 prediksi bukan | prediksi spam")
print(f"label bukan      {matriks[0][0]:14} | {matriks[0][1]}")
print(f"label spam       {matriks[1][0]:14} | {matriks[1][1]}")
```

## Kata yang Dipelajari Model

Daftar berikut menampilkan kata dengan probabilitas tertinggi pada masing-masing kelas menurut model Naive Bayes.

```python
nama_kata = vector.get_feature_names_out()
for indeks, kategori in enumerate(model.classes_):
    kata_teratas = model.feature_log_prob_[indeks].argsort()[-10:][::-1]
    print(f"{kategori.upper()}: {', '.join(nama_kata[kata_teratas])}")
```
