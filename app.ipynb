# PROJEK BELAJAR MACHINE LEARNING - DETEKSI SPAM

```python
# DETEKSI SPAM
%pip install scikit-learn -q

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score

#DATA - Semakin banyak data, semakin pintar
chat = [
    # SPAM
    "selamat anda menang hadiah 10 juta klik link ini",
    "pinjaman online cair cepat tanpa jaminan bunga rendah",
    "gratis iphone 15 cuma isi data diri anda",
    "akun anda terblokir verifikasi sekarang klik disini",
    "dapatkan saldo dana gratis 500rb",
    "lowongan kerja gaji 15jt tanpa pengalaman hubungi wa ini",
    "anda dapat voucher belanja gratis",
    "transfer sekarang untuk klaim hadiah anda",
    "promo pulsa murah 100rb cuma bayar 10rb",
    "selamat nomor anda terpilih dapat motor gratis",
    "investasi modal 100rb untung 10jt per hari",
    "cek bansos anda cair klik link berikut",
    "kirim kode otp anda untuk verifikasi hadiah",
    "diskon 90% semua produk hanya hari ini",
    "anda menang undian berhadiah mobil",

    # BUKAN SPAM
    "bro besok jadi futsal ga jam 4",
    "tugas mtk udah dikumpul belum",
    "otw rumah lu 10 menit lagi",
    "bu nanti jemput jam berapa",
    "pinjem duit 20k dong besok gue ganti",
    "eh film yang kemarin seru banget",
    "besok bawa buku paket ya jangan lupa",
    "makan siang dimana kita",
    "nilai ulangan udah keluar cek di grup",
    "nanti malem mabar ga",
    "kirim foto catetan yang tadi dong",
    "hujan deres disini lu udah pulang belum",
    "besok ada upacara pake baju putih",
    "gue udah di depan gerbang",
    "jangan lupa bawa bola",
]

label = ["spam"]*15 + ["bukan"]*15

#BAGI DATA - 80% buat belajar, 20% buat ujian
x_train, x_test, y_train, y_test = train_test_split(chat, label, test_size=0.2, random_state=42)

#UBAH KATA JADI ANGKA
vector = TfidfVectorizer()
x_train_angka = vector.fit_transform(x_train)
x_test_angka = vector.transform(x_test)

#LATIH MODEL
model = MultinomialNB()
model.fit(x_train_angka, y_train)

#CEK AKURASI
hasil = model.predict(x_test_angka)
print(f"Akurasi model lu: {accuracy_score(y_test, hasil)*100:.0f}%")
print("---")

#COBA SENDIRI - Ganti-ganti isi chat disini
def cek_spam(teks):
    angka = vector.transform([teks])
    prediksi = model.predict(angka)[0]
    prob = model.predict_proba(angka)[0].max() * 100
    print(f"Chat: {teks}")
    print(f"Hasil: {prediksi.upper()} - yakin {prob:.0f}%")
    print("---")

cek_spam("SELAMAT ANDA MENANG HADIAH GRATIS KLIK LINK")
cek_spam("eh besok bawa buku paket ya jangan lupa")
cek_spam("assalamualaikum manteman")
cek_spam("Dilms ya, kalau disini saya gk tanggung jawab kalau tidak ternilai, karena bisa hilang fileny.")
```
